and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
from openai import OpenAI
from mistralai.client import Mistral as MistralClient
from IPython.display import Markdown, display, update_display

import os
import mistralai
import ollama
import dotenv
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
MODEL_MISTRAL = 'mistral-large-latest'
```

```python
# set up environment
dotenv.load_dotenv(override=True)
gpt_key = os.environ["OPENAI_API_KEY"]
mistral_key = os.environ["MISTRAL_API_KEY"]

openai = OpenAI()
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
messages=[{"role": "user", "content": question}]
```

```python
# Get gpt-4o-mini to answer, with streaming
openAIStream = openai.chat.completions.create(model=MODEL_GPT, messages=messages, stream=True)

response = ""

display_handle = display(Markdown(""), display_id=True)
for chunk in openAIStream:
    response += chunk.choices[0].delta.content or ""
    update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
local_model_name = os.getenv("LOCAL_OLLAMA_MODEL_NAME")
response_stream = ollama.generate(model=local_model_name, prompt=question, stream=True)

response = ""
display_handle = display(Markdown(""), display_id=True)

for chunk in response_stream:
    response += chunk["response"] or ""
    update_display(Markdown(response), display_id=display_handle.display_id)
```
